```python
import os
import shutil 
from glob import glob
```

```python
DATA_BASE_PATH = r'C:\Users\kameda\Documents\projects\dataset'
OUTPUT_BASE_DIR = r'C:\Users\kameda\Documents\projects\SwinUNETR\dataset'
PROJECT_NAME = r'practice_40cases'
```

```python
# pathの作成
img_dir_name = os.listdir(os.path.join(DATA_BASE_PATH, 'half_cropped_20240609'))
print(img_dir_name)
print(len(img_dir_name))
img_paths = [os.path.join(DATA_BASE_PATH, 'half_cropped_20240609', dir, 'raw_mri_cropped.mhd') for dir in img_dir_name if dir.rsplit('_',1)[1] != 'error']
img_raw_paths = [os.path.join(DATA_BASE_PATH, 'half_cropped_20240609', dir, 'raw_mri_cropped.raw') for dir in img_dir_name if dir.rsplit('_',1)[1] != 'error']
print(img_paths)
print(len(img_paths))
lm_paths = [os.path.join(DATA_BASE_PATH, 'Kono_landmarks', dir, 'landmark_resorted.fcsv') for dir in img_dir_name]
print(lm_paths)
```

```python
output_base_path = os.path.join(OUTPUT_BASE_DIR, PROJECT_NAME)
os.makedirs(output_base_path, exist_ok=True)
for id, img_path, img_raw_path, lm_path in zip(img_dir_name, img_paths, img_raw_paths, lm_paths):
    output_input_id_path = os.path.join(output_base_path, 'input', id)
    os.makedirs(output_input_id_path, exist_ok=True)
    output_fcsv_id_path = os.path.join(output_base_path, 'fcsv', id)
    os.makedirs(output_fcsv_id_path, exist_ok=True)
    shutil.copy(img_path, os.path.join(output_input_id_path, 'raw.mhd'))
    shutil.copy(img_raw_path, os.path.join(output_input_id_path, 'raw.raw'))
    shutil.copy(lm_path, os.path.join(output_fcsv_id_path, 'landmark.fcsv'))
```
